```python

import streamlit as st

def start_quiz():

    st.title("🇵🇰 Welcome to Pakistan Quiz!")

    que_1 = st.text_input("1. What is the name of Pakistan's national animal?")
    que_2 = st.text_input("2. Who was the first Prime Minister of Pakistan?")
    que_3 = st.text_input("3. Who was the founder of Pakistan?")
    que_4 = st.text_input("4. Who was the national poet of Pakistan?")
    que_5 = st.text_input("5. What is the national language of Pakistan?")

    if st.button("Submit Quiz"):

        points = 0

        if que_1.lower().strip() == "markhor":
            st.success("Correct answer! 10 points gained.")
            points += 10
        else:
            st.error("Wrong answer!")

        if que_2.lower().strip() == "liaquat ali khan":
            st.success("Correct answer! 10 points gained.")
            points += 10
        else:
            st.error("Wrong answer!")

        if que_3.lower().strip() == "quaid e azam muhammad ali jinnah":
            st.success("Correct answer! 10 points gained.")
            points += 10
        else:
            st.error("Wrong answer!")

        if que_4.lower().strip() == "allama muhammad iqbal":
            st.success("Correct answer! 10 points gained.")
            points += 10
        else:
            st.error("Wrong answer!")

        if que_5.lower().strip() == "urdu":
            st.success("Correct answer! 10 points gained.")
            points += 10
        else:
            st.error("Wrong answer!")

        st.divider()

        st.subheader("🏆 Total Score")
        st.write("Your total score is:", points, "/50")

        if points >= 30:
            st.success("CONGRATULATIONS! YOU HAVE PASSED THE TEST 🎉")
        else:
            st.warning("SORRY, TRY AGAIN NEXT TIME!")

start_quiz()
```
